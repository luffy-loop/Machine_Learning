```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create the input data (X)

# There are 5 data points (N = 5)
# Each point has 2 input features (L = 2)

X = np.array([
    [1, 2],
    [2, 1],
    [3, 4],
    [4, 3],
    [5, 5]
])
```

```python
# Create the output values (Target)

# There is one output for each input (M = 1)

y = np.array([5, 6, 9, 10, 13])
```

```python
# Add the bias (intercept) term

# np.ones(len(X)) creates a column of ones
# np.c_ joins this column with X

X = np.c_[np.ones(len(X)), X]
```

```python
# Standard Least Squares Formula

# W = (XᵀX)^(-1) XᵀY

w = np.linalg.inv(X.T @ X) @ X.T @ y
```

```python
# Predict output values

# y_pred = XW

y_pred = X @ w
```

```python
# Display the optimal weights

print("Optimal Weights:", w)
```

```python
# Plot Actual Data

plt.scatter(range(len(y)), y, color='blue', label='Actual')
```

```python
# Plot Predicted Model

plt.plot(range(len(y)), y_pred, color='red', label='Least Squares Model')

plt.xlabel("Data Point")
plt.ylabel("Output")

plt.legend()
plt.show()
```
